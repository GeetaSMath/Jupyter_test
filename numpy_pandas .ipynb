# find out monthly sales analysis  min,max,average

```python
###### \
import numpy as np

# Month names
months = np.array([
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
])

sales = []

print("Enter the sales for each month:")
for month in months:
    value = float(input(f"{month}: "))
    sales.append(value)

# Convert list to NumPy array
sales = np.array(sales)

print("\n--- Company Sales Analysis ---")

# Calculations
total_sales = np.sum(sales)
average_sales = np.mean(sales)
highest_sales = np.max(sales)
lowest_sales = np.min(sales)

best_month = months[np.argmax(sales)]
worst_month = months[np.argmin(sales)]

above_avg_months = months[sales > average_sales]
below_avg_months = months[sales < average_sales]

# Output
print(f"Total sales of the year: {total_sales} $")
print(f"Average monthly sales: {average_sales} $")
print(f"Highest sales: {highest_sales} $")
print(f"Lowest sales: {lowest_sales} $")
print(f"Best month: {best_month}")
print(f"Worst month: {worst_month}")
print(f"Above average months: {above_avg_months}")
print(f"Below average months: {below_avg_months}")
```

```python
import pandas as pd
data={
    "Name":['Arun','Varun','Karun','Narun','Marun'],
    "Age":[28,34,22,34,28],
    "Salary":[10000,20000,30000,52000,480000]
}
df=pd.DataFrame(data)
grouped=df.groupby("Age")["Salary"].sum()
print(grouped)
grouped_name=df.groupby(["Age","Name"])["Salary"].sum()
print(grouped_name)
```

# merging 


```python
import pandas as pd
df_customers=pd.DataFrame({
    'CustomerId':[1,2,3],
    'Name':['Ramesh','Suresh','Kalpesh']
})

df_orders=pd.DataFrame({
    'CustomerId':[1,2,3],
    'OrderAmount':[250,450,350]
})

df_merged=pd.merge(df_customers,df_orders,on="CustomerId",how='inner')
print('inner join')
print(df_merged)
```

```python
#🔹 Example (Row-wise concat)
import pandas as pd

df1 = pd.DataFrame({
    "ID": [1, 2],
    "Name": ["Ram", "Shyam"]
})

df2 = pd.DataFrame({
    "ID": [3, 4],
    "Name": ["Geeta", "Simran"]
})

result = pd.concat([df1, df2], ignore_index=True)
print("row wise concat")
print(result)

#🔹 Column-wise concat
print("column wise concat")
result = pd.concat([df1, df2], axis=1)
print(result)
```

### Merging in Pandas (pd.merge)

```python
emp = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ram", "Shyam", "Geeta"]
})

salary = pd.DataFrame({
    "ID": [1, 2, 4],
    "Salary": [50000, 60000, 70000]
})

merged = pd.merge(emp, salary, on="ID", how="inner")
print(merged)
```

```python
"""| Type    | Description               |
| ------- | ------------------------- |
| `inner` | Only common rows          |
| `left`  | All left + matching right |
| `right` | All right + matching left |
| `outer` | All rows                  |"""
```
